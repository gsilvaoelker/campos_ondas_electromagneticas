## 2. De dónde sale todo esto

### 2.1 Un solo número complejo lo describe todo

En un medio sin pérdidas la onda solo acumula fase: $e^{-j\beta z}$. Si el
medio conduce, además pierde amplitud. Ambos efectos caben en un único
número complejo:

$$
e^{-\gamma z} = e^{-\alpha z}\,e^{-j\beta z},
\qquad \gamma = \alpha + j\beta .
$$

La parte real $\alpha$ apaga la onda; la parte imaginaria $\beta$ la hace
avanzar. Cuando $\sigma = 0$, $\gamma$ es imaginario puro y recuperamos el
caso de la semana 7.

### 2.2 Qué tan malo es el medio: la tangente de pérdidas

El criterio para decidir si un material es "buen conductor" o "buen
dieléctrico" es comparar la corriente de conducción con la de
desplazamiento. Ese cociente es la tangente de pérdidas.

Un material no es conductor o aislante por naturaleza: **depende de la
frecuencia**. El agua de mar es buen conductor en radio AM y buen
dieléctrico en microondas.

### 2.3 El efecto pelicular

En un buen conductor la onda se apaga tan rápido que solo penetra una capa
delgadísima. Esa capa se llama profundidad pelicular $\delta$, y a una
profundidad $3\delta$ ya no queda ni el 5 % del campo.

Consecuencia práctica: en alta frecuencia la corriente circula por la
superficie del conductor y el interior no sirve de nada. Por eso los cables
de radiofrecuencia se hacen huecos o trenzados.

## 3. Ecuaciones

**Caso general.** Para un medio con permitividad $\varepsilon$,
permeabilidad $\mu$ y conductividad $\sigma$:

$$
\gamma = \sqrt{j\omega\mu\,(\sigma + j\omega\varepsilon)},
\qquad
\eta = \sqrt{\frac{j\omega\mu}{\sigma + j\omega\varepsilon}} .
$$

**Tangente de pérdidas y conductividad equivalente:**

$$
\tan\delta = \frac{\sigma}{\omega\varepsilon}
\qquad\Longrightarrow\qquad
\sigma = \omega\,\varepsilon\,\tan\delta .
$$

**Buen conductor** ($\sigma \gg \omega\varepsilon$). Las expresiones se
simplifican mucho:

$$
\alpha = \beta = \sqrt{\pi f \mu \sigma},
\qquad
\delta = \frac{1}{\alpha},
\qquad
\eta = (1+j)\sqrt{\frac{\pi f\mu}{\sigma}} .
$$

**Potencia promedio (vector de Poynting):**

$$
S_{\text{prom}} = \frac{1}{2}\,|E_0|^2\,
\operatorname{Re}\!\left(\frac{1}{\eta^{*}}\right)
\qquad [\text{W}/\text{m}^2].
$$

**Velocidad de fase:**

$$
u_p = \frac{\omega}{\beta} = \frac{\omega}{\operatorname{Im}(\gamma)} .
$$

## 4. Qué significa físicamente

**En el cobre, la onda no entra.** A 1 MHz la profundidad pelicular es de
unos 66 micrómetros: menos que el grosor de una hoja de papel. Un blindaje
de cobre de un milímetro basta para dejar afuera prácticamente todo el campo.

**La fase de $\eta$ dice cuánto se desfasan $E$ y $H$.** En un buen conductor
$\eta$ tiene fase de exactamente 45°, así que $H$ se atrasa un octavo de
ciclo respecto de $E$. En un buen dieléctrico la fase es casi cero y se
recupera la situación de la semana 7.

**$|\eta|$ es diminuto en el cobre.** Vale fracciones de miliohm, frente a
los 377 $\Omega$ del vacío. Ese salto brutal de impedancia es la razón de
que un metal refleje casi toda la onda que le llega: es lo mismo que ocurre
en un espejo.

**En el dieléctrico con pérdidas la onda sí viaja, pero se va gastando.**
Con $\tan\delta = 0.02$ la onda recorre metros antes de apagarse. Es el
caso de una señal de WiFi cruzando una pared.

## 5. Parámetros modificables

Esta es la única celda que conviene editar. Cambie un valor, ejecute el notebook
completo y compare con lo que tenía antes.

```python
# --- Problema 1: buen conductor (cobre) ---
frecuencia_conductor = 1.0e6   # frecuencia [Hz]
conductividad = 5.8e7          # conductividad del cobre [S/m]
amplitud_E = 1.0               # amplitud del campo incidente [V/m]
profundidades = 3.0            # cuántas delta avanzar para la evaluación

# --- Problema 2: dieléctrico con pérdidas ---
frecuencia_dielectrico = 2.4e9  # frecuencia [Hz]
eps_r = 4.0                     # permitividad relativa
tangente_perdidas = 0.02        # tan(delta)
distancia = 1.0                 # distancia para evaluar la atenuación [m]
```

## 6. Implementación

### 6.1 Problema 1 — el cobre

```python
alpha_conductor = atenuacion_buen_conductor(frecuencia_conductor, conductividad)
delta = profundidad_de_penetracion(frecuencia_conductor, conductividad)
eta_conductor = impedancia_buen_conductor(frecuencia_conductor, conductividad)

campo_restante = amplitud_E * np.exp(-profundidades)
potencia_superficie = potencia_promedio(amplitud_E, eta_conductor)
```

### 6.2 Problema 2 — el dieléctrico con pérdidas

```python
sigma_equivalente = conductividad_equivalente(
    frecuencia_dielectrico, eps_r, tangente_perdidas
)
gamma = constante_propagacion(frecuencia_dielectrico, eps_r, sigma_equivalente)
eta_dielectrico = impedancia_intrinseca_compleja(
    frecuencia_dielectrico, eps_r, sigma_equivalente
)

atenuacion_a_distancia = np.exp(-gamma.real * distancia)
velocidad = velocidad_de_fase(frecuencia_dielectrico, gamma)
```

## 7. Resultados numéricos

```python
tabla_resultados(
    [
        ("Atenuación (= constante de fase)", "alpha = beta", alpha_conductor, "1/m"),
        ("Profundidad pelicular", "delta", delta, "m"),
        ("Profundidad pelicular", "delta", delta * 1.0e6, "micrómetros"),
        ("Impedancia, parte real", "Re(eta)", eta_conductor.real, "ohm"),
        ("Impedancia, parte imaginaria", "Im(eta)", eta_conductor.imag, "ohm"),
        ("Fase de la impedancia", "arg(eta)", np.rad2deg(np.angle(eta_conductor)), "grados"),
        (f"Campo a {profundidades:.0f} delta", "|E|", campo_restante, "V/m"),
        ("Potencia promedio en la superficie", "S_prom", potencia_superficie, "W/m^2"),
    ]
)
```

```python
tabla_resultados(
    [
        ("Conductividad equivalente", "sigma", sigma_equivalente, "S/m"),
        ("Atenuación", "alpha", gamma.real, "1/m"),
        ("Constante de fase", "beta", gamma.imag, "rad/m"),
        ("Impedancia, parte real", "Re(eta)", eta_dielectrico.real, "ohm"),
        ("Impedancia, parte imaginaria", "Im(eta)", eta_dielectrico.imag, "ohm"),
        (f"Campo restante a {distancia:.1f} m", "|E|/|E_0|", atenuacion_a_distancia, "-"),
        ("Velocidad de fase", "u_p", velocidad, "m/s"),
    ]
)
```

La velocidad obtenida debería quedar cerca de $c/\sqrt{\varepsilon_r}$: con
pérdidas pequeñas, el medio se comporta casi como uno sin pérdidas.

Para dimensionar cuánto importa conocer bien $\varepsilon_r$, comparamos
también contra un valor un 10 % mayor.

```python
velocidad_sin_perdidas = VELOCIDAD_LUZ / np.sqrt(eps_r)
eps_r_alternativo = 4.4
velocidad_alternativa = VELOCIDAD_LUZ / np.sqrt(eps_r_alternativo)

print(f"u_p con pérdidas             = {velocidad:.6e} m/s")
print(f"c / sqrt(eps_r = {eps_r:.1f})        = {velocidad_sin_perdidas:.6e} m/s")
print(f"Diferencia relativa          = {abs(velocidad / velocidad_sin_perdidas - 1.0):.3e}")
print()
print(f"c / sqrt(eps_r = {eps_r_alternativo:.1f})        = {velocidad_alternativa:.6e} m/s")
print(f"Cambio al subir eps_r un 10 %: "
      f"{abs(velocidad_alternativa / velocidad_sin_perdidas - 1.0) * 100.0:.2f} %")
```

## 8. Visualización

A la izquierda, el cobre en escala de profundidades peliculares. A la
derecha, el dieléctrico en metros. Fíjese en las escalas del eje horizontal.

```python
fig, (eje_conductor, eje_dielectrico) = plt.subplots(1, 2, figsize=(9.5, 4.0))

u = np.linspace(0.0, 6.0, 300)
eje_conductor.plot(u, np.exp(-u))
eje_conductor.axhline(np.exp(-1.0), color="black", linestyle=":", label="1/e = 36.8 %")
eje_conductor.axvline(1.0, color="black", linestyle="--")
eje_conductor.set_xlabel("Profundidad en unidades de delta")
eje_conductor.set_ylabel("|E| / |E(0)|")
eje_conductor.set_title(f"Cobre (delta = {delta * 1e6:.1f} micrómetros)")
eje_conductor.legend()

z = np.linspace(0.0, 3.0, 300)
eje_dielectrico.plot(z, np.exp(-gamma.real * z), color="tab:orange")
eje_dielectrico.scatter([distancia], [atenuacion_a_distancia], color="black",
                        zorder=5, label="distancia evaluada")
eje_dielectrico.set_xlabel("z (m)")
eje_dielectrico.set_ylabel("|E| / |E(0)|")
eje_dielectrico.set_title("Dieléctrico con pérdidas")
eje_dielectrico.legend()

fig.tight_layout()
plt.show()
```

## 9. Qué nos dicen los resultados

**Las dos curvas tienen la misma forma pero escalas incomparables.** Ambas
son exponenciales decrecientes. La diferencia es que el cobre se come la onda
en decenas de micrómetros y el dieléctrico tarda metros: una diferencia de
unos cinco órdenes de magnitud.

**A tres profundidades peliculares queda un 5 %.** Como $e^{-3} = 0.0498$,
casi toda la acción ocurre en la primera $\delta$. Ésta es la regla que se
usa para dimensionar blindajes.

**La fase de $\eta$ en el cobre es exactamente 45°.** No es aproximadamente
45: el factor $(1+j)$ lo garantiza. En el dieléctrico, en cambio, la fase es
de menos de un grado, y por eso su velocidad coincide con
$c/\sqrt{\varepsilon_r}$ hasta la cuarta cifra.

**La onda en el dieléctrico viaja a la mitad de $c$.** Con
$\varepsilon_r = 4.0$, $u_p \approx 1.5\times10^{8}$ m/s. Ese retardo es lo
que explica por qué una señal que atraviesa una pared llega desfasada
respecto de la que va por el aire.
