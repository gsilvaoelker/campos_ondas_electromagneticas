## 2. De dónde sale todo esto

### 2.1 Qué es la polarización

En una onda plana el campo eléctrico es perpendicular a la dirección de
propagación, pero dentro de ese plano puede apuntar donde quiera. La
polarización describe **cómo se mueve la punta del vector $\mathbf{E}$** en
ese plano a medida que pasa el tiempo.

Todo se reduce a superponer dos componentes perpendiculares, $E_x$ y $E_y$.
Lo que decide la forma de la trayectoria son dos cosas: la razón entre sus
amplitudes y el desfase entre ellas.

### 2.2 Los tres casos

- **Lineal:** las dos componentes están en fase (o en contrafase). La punta
  va y viene por una recta.
- **Circular:** amplitudes iguales y desfase de $90^\circ$. La punta
  describe un círculo.
- **Elíptica:** el caso general. Las dos anteriores son casos particulares
  de éste.

En este notebook usamos $E_x = E_{x0}\cos t$ y $E_y = E_{y0}\sin t$: hay
$90^\circ$ de desfase, así que la trayectoria es una elipse con los ejes
alineados con $x$ e $y$.

### 2.3 Por qué esto le importa a una pantalla

Un polarizador solo deja pasar la componente del campo paralela a su eje. Si
usted pone dos polarizadores cruzados a $90^\circ$, no pasa nada de luz.

Un píxel LCD usa cristal líquido entre polarizadores para modificar el estado
de polarización mediante un voltaje. Aquí usamos un modelo óptico idealizado en
el que esa acción equivale a un giro controlable. Un panel real requiere un
modelo electroóptico más detallado.

## 3. Ecuaciones

**Componentes del campo (desfase de $90^\circ$):**

$$
E_x(t) = E_{x0}\cos(\omega t),
\qquad
E_y(t) = E_{y0}\sin(\omega t).
$$

Eliminando el tiempo se obtiene la ecuación de una elipse:

$$
\left(\frac{E_x}{E_{x0}}\right)^{2} + \left(\frac{E_y}{E_{y0}}\right)^{2} = 1 .
$$

**Razón axial** (eje mayor dividido por eje menor):

$$
\mathrm{RA} = \frac{\max(E_{x0},\,E_{y0})}{\min(E_{x0},\,E_{y0})} \;\ge\; 1 .
$$

Vale 1 para polarización circular y tiende a infinito para lineal.

**Transmisión de un píxel LCD entre polarizadores cruzados**, cuando el
cristal líquido gira la polarización un ángulo $\phi$:

$$
\frac{I}{I_0} = \sin^{2}\phi .
$$

Con $\phi = 0$ el píxel está apagado; con $\phi = 90^\circ$, encendido al
máximo.

## 4. Qué significa físicamente

**La elipse no es una trayectoria en el espacio.** El campo no se desplaza:
lo que gira es la *dirección* del vector en un punto fijo. La onda sigue
viajando en $z$.

**El sentido de giro importa.** En nuestro caso, en $t = 0$ el campo apunta
hacia $+x$ (porque $\cos 0 = 1$ y $\sin 0 = 0$), y un instante después
aparece componente $+y$. Es decir, gira de $+x$ hacia $+y$. Invertir el signo
de una componente invierte el giro, y eso distingue una onda circular derecha
de una izquierda.

**La razón axial resume la forma en un número.** Con $E_{x0} = 4$ y
$E_{y0} = 3$ da 1.33: una elipse moderadamente achatada. Las antenas de GPS
se especifican con este número, porque una razón axial mala significa perder
señal.

**La curva de transmisión del LCD no es lineal.** Como va con $\sin^2\phi$,
los primeros grados de giro casi no aclaran el píxel, y cerca de $90^\circ$
tampoco cambia mucho. La sensibilidad máxima está en $45^\circ$. Los
controladores de pantalla tienen que corregir esta curva para que los grises
se vean parejos.

```python
# --- Problema 1: elipse de polarización ---
amplitud_x = 4.0   # amplitud de la componente x, E_x0 [V/m]
amplitud_y = 3.0   # amplitud de la componente y, E_y0 [V/m]

# --- Problema 2: píxel LCD ---
giro_grados = 20.0  # ángulo que gira el cristal líquido [grados]
```

## 6. Implementación

### 6.1 Problema 1 — la trayectoria del campo

```python
if amplitud_x < 0.0 or amplitud_y < 0.0:
    raise ValueError("Las amplitudes deben ser no negativas.")

t = np.linspace(0.0, 2.0 * np.pi, 500)
campo_x = amplitud_x * np.cos(t)
campo_y = amplitud_y * np.sin(t)

eje_mayor = max(amplitud_x, amplitud_y)
eje_menor = min(amplitud_x, amplitud_y)
# Si un semieje es nulo la polarización es lineal y la razón axial diverge.
razon_axial = np.inf if eje_menor == 0.0 else eje_mayor / eje_menor

# Sentido de giro: comparamos el instante inicial con uno inmediatamente posterior.
hay_giro = amplitud_x > 0.0 and amplitud_y > 0.0
```

### 6.2 Problema 2 — el píxel LCD

```python
transmision = np.sin(np.deg2rad(giro_grados)) ** 2
transmision_maxima = np.sin(np.deg2rad(90.0)) ** 2
```

## 7. Resultados numéricos

```python
tabla_resultados(
    [
        ("Amplitud en x", "E_x0", amplitud_x, "V/m"),
        ("Amplitud en y", "E_y0", amplitud_y, "V/m"),
        ("Razón axial", "RA", razon_axial, "-"),
        ("Semieje mayor", "a", max(amplitud_x, amplitud_y), "V/m"),
        ("Semieje menor", "b", min(amplitud_x, amplitud_y), "V/m"),
    ]
)
```

```python
if np.isclose(razon_axial, 1.0):
    tipo = "circular"
elif razon_axial >= 50.0:
    tipo = "lineal (o prácticamente lineal)"
else:
    tipo = "elíptica"

print(f"Razón axial: {razon_axial:.6g}")
print(f"Tipo de polarización: {tipo}")
if hay_giro:
    print("Desde t = 0 la punta se mueve de +x hacia +y.")
else:
    print("No hay giro: el campo oscila sobre una sola recta.")
```

```python
tabla_resultados(
    [
        (f"Transmisión con giro de {giro_grados:.1f} grados", "I/I_0", transmision, "-"),
        ("Transmisión con giro de 90 grados", "I/I_0", transmision_maxima, "-"),
        ("Brillo relativo del píxel", "porcentaje", transmision * 100.0, "%"),
    ]
)
```

## 8. Visualización

A la izquierda, la trayectoria del campo. A la derecha, cuánto se enciende el
píxel según cuánto gire el cristal líquido.

```python
fig, (eje_elipse, eje_lcd) = plt.subplots(1, 2, figsize=(9.5, 4.0))

eje_elipse.plot(campo_x, campo_y)
eje_elipse.scatter([campo_x[0]], [campo_y[0]], color="black", zorder=5, label="t = 0")
eje_elipse.annotate("", xy=(campo_x[25], campo_y[25]), xytext=(campo_x[0], campo_y[0]),
                    arrowprops=dict(arrowstyle="->", color="tab:red", lw=2))
eje_elipse.set_aspect("equal", adjustable="box")
eje_elipse.set_xlabel("E_x (V/m)")
eje_elipse.set_ylabel("E_y (V/m)")
eje_elipse.set_title(f"Elipse de polarización (RA = {razon_axial:.2f})")
eje_elipse.legend()

angulos = np.linspace(0.0, 90.0, 361)
eje_lcd.plot(angulos, np.sin(np.deg2rad(angulos)) ** 2)
eje_lcd.scatter([giro_grados], [transmision], color="black", zorder=5,
                label="giro elegido")
eje_lcd.set_xlabel("Giro del cristal líquido (grados)")
eje_lcd.set_ylabel("I / I_0")
eje_lcd.set_title("Píxel entre polarizadores cruzados")
eje_lcd.legend()

fig.tight_layout()
plt.show()
```

## 9. Qué nos dicen los resultados

**La elipse está achatada en $y$.** Como $E_{x0} > E_{y0}$, el eje mayor
queda sobre $x$. La flecha roja indica el sentido de giro: de $+x$ hacia
$+y$.

**La razón axial también puede expresarse en decibelios.** Para este caso,
$20\log_{10}(1.33) \approx 2.5$ dB. Decidir si ese valor es aceptable requiere
compararlo con la especificación de la aplicación; el notebook no fija ese
criterio.

**Con 20° de giro el píxel apenas se enciende.** La transmisión es de
aproximadamente 0.117, es decir un 12 % del brillo máximo: el píxel se ve
gris oscuro. Para llegar a la mitad del brillo hace falta girar 45°.

**La curva es monótona entre 0° y 90°.** Los ángulos complementarios no dan
el mismo brillo: $\sin^2(20^\circ) \approx 0.117$ y
$\sin^2(70^\circ) \approx 0.883$. Sus transmisiones suman uno.
