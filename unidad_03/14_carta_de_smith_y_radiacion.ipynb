## 2. De dónde sale todo esto

### 2.1 Qué es realmente la carta de Smith

La carta de Smith parece complicada, pero es una sola idea: **dibujar
$\Gamma$ en el plano complejo**.

Como $|\Gamma| \le 1$ para cualquier carga pasiva, todos los casos posibles
caben dentro de un círculo de radio 1. Encima de ese círculo se dibujan las
curvas que corresponden a resistencia y reactancia normalizadas constantes,
y así se puede leer la impedancia sin calcular nada.

Antes de los computadores era una herramienta de cálculo indispensable. Hoy
se sigue usando porque permite *ver* de un vistazo qué tan adaptada está una
carga y hacia dónde hay que moverse para adaptarla.

### 2.2 Normalizar

Todo en la carta va normalizado a $Z_0$: se trabaja con $z_L = Z_L/Z_0$. Así
la misma carta sirve para líneas de 50, 75 o cualquier otra impedancia.

El centro de la carta es $z = 1$, es decir, adaptación perfecta.

### 2.3 El dipolo hertziano

Es la antena más simple imaginable: un tramo de corriente mucho más corto que
la longitud de onda. No es realizable tal cual, pero cualquier antena real se
puede pensar como una suma de dipolos hertzianos, así que entenderlo es
entender el ladrillo básico de la radiación.

Lo importante es que sus campos lejanos decaen como $1/r$, no como $1/r^2$.
Por eso la energía radiada llega lejos: la potencia cae como $1/r^2$, pero
eso es solo la dilución geométrica sobre una esfera.

## 3. Ecuaciones

**Impedancia normalizada y reflexión:**

$$
z_L = \frac{Z_L}{Z_0},
\qquad
\Gamma = \frac{z_L - 1}{z_L + 1},
\qquad
\mathrm{ROE} = \frac{1+|\Gamma|}{1-|\Gamma|}.
$$

**Curvas de la carta.** Los lugares de resistencia normalizada constante y de
reactancia normalizada constante son circunferencias en el plano $\Gamma$:

$$
\Gamma(r, x) = \frac{r + jx - 1}{r + jx + 1}.
$$

**Campos lejanos de un dipolo hertziano** de longitud $dl \ll \lambda$
recorrido por una corriente $I_0$:

$$
|E_\theta| = \frac{\eta_0\,\beta\,I_0\,dl\,\sin\theta}{4\pi r},
\qquad
|H_\phi| = \frac{|E_\theta|}{\eta_0},
\qquad
S = \frac{|E_\theta|^{2}}{2\eta_0}.
$$

**Patrón de radiación normalizado:**

$$
U(\theta) = \sin^{2}\theta .
$$

Máximo en el ecuador ($\theta = 90^\circ$) y cero en el eje del dipolo.

## 4. Qué significa físicamente

**El centro de la carta es la meta.** Adaptar consiste literalmente en mover
el punto hacia el centro. Cuanto más lejos del centro esté la carga, peor es
la ROE.

**Moverse por la línea es girar en la carta.** Alejarse de la carga hace girar
el punto sobre un círculo de $|\Gamma|$ constante. Una vuelta completa
corresponde a media longitud de onda, que es la periodicidad que vimos en la
semana 12.

**El dipolo no radia en la dirección en que apunta.** El $\sin\theta$ se anula
en $\theta = 0$ y $180^\circ$. Por eso, si usted se para justo bajo una
antena vertical, la señal es pésima: está en el nulo.

**Ejemplo a 915 MHz.** Los parámetros ilustran la escala del campo lejano de
un elemento hertziano impuesto. No modelan por sí solos una etiqueta RFID: para
eso harían falta la alimentación, la adaptación, la antena receptora y el
enlace de ida y vuelta.

```python
# --- Problema 1: carga en la carta de Smith ---
Z0 = 50.0            # impedancia característica [ohm]
ZL = 30.0 - 40.0j    # impedancia de la carga [ohm]

# --- Problema 2: elemento hertziano ideal ---
frecuencia = 915.0e6      # frecuencia [Hz]
corriente = 20.0e-3       # corriente en la antena I_0 [A]
longitud_dipolo = 0.03    # longitud del dipolo dl [m]
distancia = 10.0          # distancia de observación r [m]
angulo_grados = 90.0      # ángulo respecto del eje del dipolo [grados]
```

## 6. Implementación

### 6.1 Problema 1 — ubicar la carga

```python
zL = ZL / Z0
Gamma = coeficiente_reflexion(ZL, Z0)
roe = razon_onda_estacionaria(Gamma)
```

### 6.2 Problema 2 — campos del dipolo

```python
longitud_onda = longitud_de_onda(frecuencia)
beta = numero_de_onda(frecuencia)
angulo = np.deg2rad(angulo_grados)

campo_E = campo_dipolo_hertziano(
    corriente, longitud_dipolo, distancia, angulo, beta
)
campo_H = campo_E / IMPEDANCIA_VACIO
densidad_potencia = campo_E**2 / (2.0 * IMPEDANCIA_VACIO)

fraccion_de_lambda = longitud_dipolo / longitud_onda
```

## 7. Resultados numéricos

```python
tabla_resultados(
    [
        ("Resistencia normalizada", "Re(z_L)", zL.real, "-"),
        ("Reactancia normalizada", "Im(z_L)", zL.imag, "-"),
        ("Reflexión, parte real", "Re(Gamma)", Gamma.real, "-"),
        ("Reflexión, parte imaginaria", "Im(Gamma)", Gamma.imag, "-"),
        ("Módulo de la reflexión", "|Gamma|", abs(Gamma), "-"),
        ("Fase de la reflexión", "arg(Gamma)", np.rad2deg(np.angle(Gamma)), "grados"),
        ("Razón de onda estacionaria", "ROE", roe, "-"),
    ]
)
```

```python
tabla_resultados(
    [
        ("Longitud de onda", "lambda", longitud_onda, "m"),
        ("Constante de fase", "beta", beta, "rad/m"),
        ("Tamaño del dipolo", "dl/lambda", fraccion_de_lambda, "-"),
        ("Campo eléctrico", "|E_theta|", campo_E, "V/m"),
        ("Campo magnético", "|H_phi|", campo_H, "A/m"),
        ("Densidad de potencia", "S", densidad_potencia, "W/m^2"),
    ]
)
```

```python
print(f"dl / lambda = {fraccion_de_lambda:.4f}")
if fraccion_de_lambda < 0.05:
    print("dl/lambda es pequeño: la aproximación hertziana es razonable.")
else:
    print("Cuidado: dl/lambda no es mucho menor que 1; el resultado es solo orientativo.")
```

## 8. Visualización

A la izquierda, la carta de Smith con la carga marcada. A la derecha, el
patrón de radiación del dipolo.

```python
figura = plt.figure(figsize=(9.5, 4.3))

# --- Carta de Smith ---
eje_smith = figura.add_subplot(1, 2, 1)
eje_smith.add_patch(plt.Circle((0.0, 0.0), 1.0, fill=False, color="black"))

reactancias = np.linspace(-10.0, 10.0, 1200)
for resistencia in (0.0, 0.2, 0.5, 1.0, 2.0, 5.0):
    curva = (resistencia + 1.0j * reactancias - 1.0) / (resistencia + 1.0j * reactancias + 1.0)
    eje_smith.plot(curva.real, curva.imag, color="0.75", linewidth=0.7)

resistencias = np.linspace(0.0, 10.0, 1000)
for reactancia in (-5.0, -2.0, -1.0, -0.5, 0.5, 1.0, 2.0, 5.0):
    curva = (resistencias + 1.0j * reactancia - 1.0) / (resistencias + 1.0j * reactancia + 1.0)
    eje_smith.plot(curva.real, curva.imag, color="0.82", linewidth=0.7)

eje_smith.add_patch(plt.Circle((0.0, 0.0), abs(Gamma), fill=False,
                               linestyle="--", color="tab:blue"))
eje_smith.scatter([Gamma.real], [Gamma.imag], color="tab:red", zorder=5, label="carga")
eje_smith.scatter([0.0], [0.0], color="black", marker="+", s=80, label="adaptación")
eje_smith.set_aspect("equal", adjustable="box")
eje_smith.set_xlim(-1.05, 1.05)
eje_smith.set_ylim(-1.05, 1.05)
eje_smith.set_xlabel("Re(Gamma)")
eje_smith.set_ylabel("Im(Gamma)")
eje_smith.set_title("Carta de Smith simplificada")
eje_smith.grid(False)
eje_smith.legend(fontsize=8, loc="lower left")

# --- Patrón de radiación ---
eje_patron = figura.add_subplot(1, 2, 2, projection="polar")
theta = np.linspace(0.0, 2.0 * np.pi, 720)
eje_patron.plot(theta, np.sin(theta) ** 2)
eje_patron.set_title("Potencia normalizada del dipolo")

figura.tight_layout()
plt.show()
```

## 9. Qué nos dicen los resultados

**La carga está en la mitad inferior de la carta.** Como su reactancia es
negativa ($-40j$), es capacitiva, y en la carta de Smith eso significa quedar
bajo el eje horizontal. Las cargas inductivas quedan arriba.

**La ROE se lee como distancia al centro.** El círculo azul punteado tiene
radio $|\Gamma| = 0.5$ exacto, y la ROE resulta exactamente 3.0. Cuanto más
grande ese círculo, peor la adaptación; el centro marcado con la cruz es el
objetivo. Con $|\Gamma| = 0.5$ vuelve un 25 % de la potencia.

**El elemento no es extremadamente corto.** Con $dl/\lambda \approx 0.09$,
la condición $dl\ll\lambda$ es marginal. El código lo advierte: el resultado
sirve para ver tendencias, no como validación precisa de una antena real.

**El patrón tiene forma de ocho.** Los dos lóbulos apuntan perpendicular al
dipolo, y hay ceros perfectos en el eje. En tres dimensiones esto es un
toroide, una dona con el dipolo atravesándola por el agujero.

**El campo calculado no establece un alcance de comunicación.** Un alcance
requiere potencia disponible, ganancias, polarización, adaptación, sensibilidad
del receptor y, en retrodispersión, el enlace de regreso. Este notebook solo
calcula el campo del elemento ideal impuesto.
