## 2. De dónde sale todo esto

### 2.1 Del campo al potencial

El potencial se obtiene integrando el campo desde el infinito, donde se toma
$V = 0$. Para la esfera hay que hacerlo por tramos: primero desde $\infty$
hasta $a$ con la fórmula exterior, y después desde $a$ hasta $r$ con la
interior.

El resultado es **continuo** en $r = a$ aunque el campo tenga ahí un quiebre.
Es general: $V$ es la integral de $E$, y la integral de una función con un
quiebre no salta.

### 2.2 Por qué hacía falta la corriente de desplazamiento

Imagine un condensador conectado a una fuente alterna. Tome un contorno
alrededor del cable y pregunte qué corriente lo atraviesa.

Si elige una superficie que corta el cable, la respuesta es la corriente de
conducción. Si elige una que pasa entre las placas, la respuesta sería cero,
porque ahí no circula ninguna carga. Dos respuestas distintas para el mismo
contorno: eso es una contradicción.

El término $\partial\mathbf{D}/\partial t$ la resuelve. Entre las placas no
hay carga moviéndose, pero sí un campo eléctrico que cambia, y eso cuenta
como corriente.

## 3. Ecuaciones

**Problema 1 — esfera de radio $a$ con carga uniforme:**

$$
Q = \rho_v \frac{4}{3}\pi a^3,
\qquad
V(r) =
\begin{cases}
\dfrac{\rho_v\left(3a^2 - r^2\right)}{6\varepsilon_0}, & r < a, \\[2ex]
\dfrac{Q}{4\pi\varepsilon_0 r}, & r \ge a .
\end{cases}
$$

Dos valores que conviene recordar:

$$
V(a) = \frac{Q}{4\pi\varepsilon_0 a} = \frac{\rho_v a^2}{3\varepsilon_0},
\qquad
V(0) = \frac{\rho_v a^2}{2\varepsilon_0} = \frac{3}{2}\,V(a).
$$

**Problema 2 — condensador con señal alterna:**

$$
E_0 = \frac{V_0}{d},
\qquad
\omega = 2\pi f,
\qquad
\mathbf{J}_d = \frac{\partial\mathbf{D}}{\partial t}.
$$

Para $E(t) = E_0\cos\omega t$, la amplitud de la corriente de desplazamiento
total sobre el área $\pi R^2$ es

$$
I_{d0} = \varepsilon_0\,\omega\,E_0\,\pi R^2,
\qquad
H_\phi(R) = \frac{I_{d0}}{2\pi R}.
$$

## 4. Qué significa físicamente

**$E(0) = 0$ no implica $V(0) = 0$.** Es el error más frecuente en este tema.
El campo se anula en el centro por simetría, pero el potencial ahí es
**máximo**: llevar una carga desde el infinito hasta el centro cuesta trabajo
durante todo el trayecto. El valor $V(0) = \tfrac{3}{2}V(a)$ vale la pena
memorizarlo.

**La corriente de desplazamiento no transporta carga.** Entre las placas no se
mueve ninguna partícula. Lo único que pasa es que el campo cambia. Y sin
embargo produce campo magnético igual que una corriente de verdad. Esa
simetría entre $\mathbf{E}$ y $\mathbf{H}$ es lo que hace posibles las ondas
de la Unidad 2.

**A más frecuencia, más corriente de desplazamiento.** El factor $\omega$
explica por qué un condensador bloquea la corriente continua y en cambio
conduce cada vez mejor a frecuencias altas.

## 5. Parámetros modificables

Esta es la única celda que conviene editar. Cambie un valor, ejecute el notebook
completo y compare con lo que tenía antes.

```python
# --- Problema 1: esfera con carga uniforme ---
radio_esfera = 0.03    # radio a [m]
rho_v = 1.0e-6         # densidad de carga volumétrica [C/m^3]

# --- Problema 2: condensador con señal alterna ---
radio_placas = 0.03    # radio de las placas R [m]
separacion = 0.001     # separación entre placas d [m]
voltaje_pico = 20.0    # amplitud del voltaje V_0 [V]
frecuencia = 1.0e6     # frecuencia f [Hz]
```

## 6. Implementación

### 6.1 Problema 1 — el potencial de la esfera

```python
carga_total = carga_total_esfera_uniforme(radio_esfera, rho_v)
V_superficie = float(potencial_esfera_uniforme(radio_esfera, radio_esfera, rho_v))
V_centro = float(potencial_esfera_uniforme(0.0, radio_esfera, rho_v))
```

### 6.2 Problema 2 — la corriente de desplazamiento

```python
omega = 2.0 * np.pi * frecuencia
campo_pico = voltaje_pico / separacion
corriente_desplazamiento = EPSILON_0 * omega * campo_pico * np.pi * radio_placas**2
campo_H_borde = corriente_desplazamiento / (2.0 * np.pi * radio_placas)
```

## 7. Resultados numéricos

```python
tabla_resultados(
    [
        ("Carga total", "Q", carga_total, "C"),
        ("Potencial en la superficie", "V(a)", V_superficie, "V"),
        ("Potencial en el centro", "V(0)", V_centro, "V"),
        ("Razón entre ambos", "V(0)/V(a)", V_centro / V_superficie, "-"),
    ]
)
```

```python
tabla_resultados(
    [
        ("Frecuencia angular", "omega", omega, "rad/s"),
        ("Campo eléctrico pico", "E_0", campo_pico, "V/m"),
        ("Corriente de desplazamiento", "I_d0", corriente_desplazamiento, "A"),
        ("Campo magnético en el borde", "H_phi(R)", campo_H_borde, "A/m"),
    ]
)
```

Comprobación de $V(0)$ contra la fórmula cerrada
$\rho_v a^2 / (2\varepsilon_0)$:

```python
V_centro_cerrado = rho_v * radio_esfera**2 / (2.0 * EPSILON_0)
print(f"V(0) numérico  = {V_centro:.6f} V")
print(f"V(0) cerrado   = {V_centro_cerrado:.6f} V")
print(f"Diferencia     = {abs(V_centro - V_centro_cerrado):.3e} V")
```

## 8. Visualización

Campo y potencial juntos. Fíjese en el quiebre de $|\mathbf{E}|$ en $r = a$ y
en que $V$ pasa por ahí sin quebrarse.

```python
r = np.linspace(1.0e-5, 3.0 * radio_esfera, 500)
campo = campo_esfera_uniforme(r, radio_esfera, rho_v)
potencial = potencial_esfera_uniforme(r, radio_esfera, rho_v)

fig, (eje_campo, eje_potencial) = plt.subplots(2, 1, sharex=True, figsize=(7.0, 6.0))

eje_campo.plot(r * 100.0, campo)
eje_campo.axvline(radio_esfera * 100.0, color="black", linestyle="--", label="superficie")
eje_campo.set_ylabel("|E| (V/m)")
eje_campo.set_title("Campo y potencial de una esfera cargada")
eje_campo.legend()

eje_potencial.plot(r * 100.0, potencial, color="tab:orange")
eje_potencial.axvline(radio_esfera * 100.0, color="black", linestyle="--")
eje_potencial.scatter([0.0, radio_esfera * 100.0], [V_centro, V_superficie],
                      color="black", zorder=5)
eje_potencial.set_xlabel("r (cm)")
eje_potencial.set_ylabel("V (V)")

fig.tight_layout()
plt.show()
```

## 9. Qué nos dicen los resultados

**$E$ se quiebra, $V$ no.** En $r = a$ la pendiente del potencial cambia de
golpe, pero el potencial mismo es continuo. Es coherente con
$\mathbf{E} = -\nabla V$: el campo **es** la pendiente, y una pendiente puede
saltar sin que la función salte.

**El potencial es máximo justo donde el campo es cero.** En el centro,
$E = 0$ y $V$ alcanza su valor más alto, un 50 % mayor que en la superficie.
Es lo mismo que pasa en el máximo de cualquier función: la derivada se anula.

**El campo magnético del condensador es real y medible.** Con 20 V a 1 MHz se
obtiene $H_\phi \approx 1.7\times10^{-2}$ A/m en el borde. No es un artificio:
es un campo de verdad, producido sin que ninguna carga cruce el espacio entre
las placas.

**Todo escala con la frecuencia.** Como $I_{d0} \propto \omega$, pasar de
1 MHz a 1 GHz multiplica por mil tanto la corriente de desplazamiento como el
campo magnético.
