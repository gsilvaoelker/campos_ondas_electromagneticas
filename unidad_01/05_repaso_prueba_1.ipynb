## 2. Fundamentos teóricos y matemáticos

### 2.1 Del campo al potencial

El potencial se obtiene integrando el campo desde el infinito, donde se toma
$V = 0$, hasta el punto de interés. Para la esfera con carga uniforme hay que
integrar por tramos: primero la rama exterior desde $\infty$ hasta $a$, y
luego la rama interior desde $a$ hasta $r$.

El potencial resulta **continuo** en $r = a$ aunque el campo tenga allí un
quiebre. Esto es general: $V$ es la integral de $E$, y la integral de una
función con un quiebre finito es continua.

### 2.2 Corriente de desplazamiento

El término $\partial\mathbf{D}/\partial t$ que Maxwell agregó a la ley de
Ampère resuelve una contradicción concreta. Considere un condensador
conectado a una fuente alterna y una superficie que pase entre las placas:
no la atraviesa ninguna corriente de conducción, pero sí hay campo magnético
alrededor. Sin el término de desplazamiento, la ley de Ampère daría dos
respuestas distintas según qué superficie se elija para el mismo contorno.

Con el término agregado, la corriente de desplazamiento entre las placas
iguala exactamente a la corriente de conducción en los cables, y la
ambigüedad desaparece.

## 3. Ecuaciones relevantes

**Problema 1 — esfera con carga volumétrica uniforme de radio $a$:**

$$
Q = \rho_v \frac{4}{3}\pi a^3,
$$

$$
V(r) =
\begin{cases}
\dfrac{\rho_v\left(3a^2 - r^2\right)}{6\varepsilon_0}, & r < a, \\[2ex]
\dfrac{Q}{4\pi\varepsilon_0 r}, & r \ge a .
\end{cases}
$$

Dos valores particulares:

$$
V(a) = \frac{Q}{4\pi\varepsilon_0 a} = \frac{\rho_v a^2}{3\varepsilon_0},
\qquad
V(0) = \frac{\rho_v a^2}{2\varepsilon_0} = \frac{3}{2}\,V(a).
$$

**Problema 2 — condensador con excitación armónica:**

$$
E_0 = \frac{V_0}{d},
\qquad
\omega = 2\pi f,
\qquad
\mathbf{J}_d = \frac{\partial\mathbf{D}}{\partial t}.
$$

Para $E(t) = E_0\cos\omega t$, la amplitud de la corriente de desplazamiento
total a través del área $\pi R^2$ es

$$
I_{d0} = \varepsilon_0\,\omega\,E_0\,\pi R^2 .
$$

Aplicando la ley de Ampère a un contorno circular de radio $R$ en el plano
medio entre las placas:

$$
H_\phi(R) = \frac{I_{d0}}{2\pi R}.
$$

## 4. Explicación física

**El potencial en el centro no es cero.** Un error frecuente es confundir
$E(0) = 0$ con $V(0) = 0$. El campo se anula en el centro por simetría, pero
el potencial allí es el **máximo**: para llevar una carga de prueba desde el
infinito hasta el centro hay que hacer trabajo contra el campo durante todo
el recorrido exterior e interior. El valor $V(0) = \tfrac{3}{2}V(a)$ es un
resultado que conviene recordar.

**La corriente de desplazamiento no transporta carga.** Entre las placas no
se mueve ninguna partícula: lo que varía es el campo eléctrico. Sin embargo,
produce campo magnético exactamente igual que una corriente de conducción de
la misma magnitud. Esta es la simetría que hace posibles las ondas
electromagnéticas, tema de la Unidad 2.

**La corriente de desplazamiento crece con la frecuencia.** El factor
$\omega$ en $I_{d0}$ explica por qué a frecuencias bajas un condensador se
comporta como circuito abierto, y a frecuencias altas conduce cada vez
mejor.

## 5. Parámetros modificables

Modifique **solo** esta celda y vuelva a ejecutar el notebook completo.

```python
# --- Problema 1: esfera con carga uniforme ---
radio_esfera = 0.03    # radio a [m]
rho_v = 1.0e-6         # densidad de carga volumétrica [C/m^3]

# --- Problema 2: condensador con excitación armónica ---
radio_placas = 0.03    # radio de las placas R [m]
separacion = 0.001     # separación entre placas d [m]
voltaje_pico = 20.0    # amplitud del voltaje V_0 [V]
frecuencia = 1.0e6     # frecuencia f [Hz]
```

## 6. Implementación del código

### 6.1 Problema 1 — potencial de la esfera

```python
carga_total = carga_total_esfera_uniforme(radio_esfera, rho_v)
V_superficie = float(potencial_esfera_uniforme(radio_esfera, radio_esfera, rho_v))
V_centro = float(potencial_esfera_uniforme(0.0, radio_esfera, rho_v))
```

### 6.2 Problema 2 — corriente de desplazamiento

```python
omega = 2.0 * np.pi * frecuencia
campo_pico = voltaje_pico / separacion
corriente_desplazamiento = EPSILON_0 * omega * campo_pico * np.pi * radio_placas**2
campo_H_borde = corriente_desplazamiento / (2.0 * np.pi * radio_placas)
```

## 7. Resultados numéricos

```python
tabla_resultados(
    [
        ("Carga total", "Q", carga_total, "C"),
        ("Potencial en la superficie", "V(a)", V_superficie, "V"),
        ("Potencial en el centro", "V(0)", V_centro, "V"),
        ("Razón entre ambos", "V(0)/V(a)", V_centro / V_superficie, "-"),
    ]
)
```

```python
tabla_resultados(
    [
        ("Frecuencia angular", "omega", omega, "rad/s"),
        ("Campo eléctrico pico", "E_0", campo_pico, "V/m"),
        ("Corriente de desplazamiento", "I_d0", corriente_desplazamiento, "A"),
        ("Campo magnético en el borde", "H_phi(R)", campo_H_borde, "A/m"),
    ]
)
```

Comprobación del valor $V(0) = \tfrac{3}{2}V(a)$ frente a la expresión
cerrada $\rho_v a^2 / (2\varepsilon_0)$:

```python
V_centro_cerrado = rho_v * radio_esfera**2 / (2.0 * EPSILON_0)
print(f"V(0) numérico  = {V_centro:.6f} V")
print(f"V(0) cerrado   = {V_centro_cerrado:.6f} V")
print(f"Diferencia     = {abs(V_centro - V_centro_cerrado):.3e} V")
```

## 8. Visualización

Campo y potencial de la esfera en la misma figura. Observe el quiebre de
$|\mathbf{E}|$ en $r = a$ y la suavidad de $V$ en ese mismo punto.

```python
r = np.linspace(1.0e-5, 3.0 * radio_esfera, 500)
campo = campo_esfera_uniforme(r, radio_esfera, rho_v)
potencial = potencial_esfera_uniforme(r, radio_esfera, rho_v)

fig, (ax_campo, ax_potencial) = plt.subplots(2, 1, sharex=True, figsize=(7.0, 6.0))

ax_campo.plot(r * 100.0, campo)
ax_campo.axvline(radio_esfera * 100.0, color="black", linestyle="--", label="superficie")
ax_campo.set_ylabel("|E| (V/m)")
ax_campo.set_title("Campo y potencial de una esfera con carga uniforme")
ax_campo.legend()

ax_potencial.plot(r * 100.0, potencial, color="tab:orange")
ax_potencial.axvline(radio_esfera * 100.0, color="black", linestyle="--")
ax_potencial.scatter([0.0, radio_esfera * 100.0], [V_centro, V_superficie],
                     color="black", zorder=5)
ax_potencial.set_xlabel("r (cm)")
ax_potencial.set_ylabel("V (V)")

fig.tight_layout()
plt.show()
```

## 9. Interpretación física de los resultados

**$E$ se quiebra, $V$ no.** En $r = a$ la derivada del potencial cambia de
pendiente bruscamente, pero el potencial mismo es continuo. Esto es
consistente con $\mathbf{E} = -\nabla V$: el campo es la pendiente del
potencial, y una pendiente puede saltar sin que la función salte.

**El potencial máximo está donde el campo es nulo.** En el centro, $E = 0$
y $V$ alcanza su valor máximo, un 50 % mayor que en la superficie. Es la
misma relación que en un punto de máximo de cualquier función: la derivada
se anula.

**El campo magnético del condensador es medible.** Con 20 V a 1 MHz se
obtiene $H_\phi \approx 1.7\times10^{-2}$ A/m en el borde de las placas.
No es un artificio matemático: es un campo real, producido sin que ninguna
carga cruce el espacio entre las placas.

**Escalamiento con la frecuencia.** Como $I_{d0} \propto \omega$, subir de
1 MHz a 1 GHz multiplica por mil tanto la corriente de desplazamiento como
el campo magnético asociado.
