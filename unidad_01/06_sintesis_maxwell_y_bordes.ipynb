## 2. De dónde sale todo esto

### 2.1 Revisar las ecuaciones una por una

Las cuatro ecuaciones de Maxwell resumen todo el electromagnetismo clásico.
En vez de aceptarlas como bloque, vale la pena revisarlas de a una y
preguntarse si están completas.

El caso interesante es la ley de Ampère. La versión
$\nabla\times\mathbf{H} = \mathbf{J}$ **está incompleta**, y se puede
demostrar sin hacer ningún experimento.

Tome la divergencia de los dos lados. La divergencia de un rotacional siempre
es cero, así que quedaría $\nabla\cdot\mathbf{J} = 0$. Pero la ecuación de
continuidad dice que $\nabla\cdot\mathbf{J} = -\partial\rho_v/\partial t$, que
no es cero si la carga varía.

Contradicción. Y se arregla agregando la corriente de desplazamiento.

### 2.2 Las unidades como herramienta

Revisar unidades no es un trámite: es un diagnóstico. Si los dos lados de una
ecuación no tienen las mismas unidades, la ecuación está mal, y usted lo sabe
sin revisar la física.

### 2.3 Comprobar con números

La continuidad de $D_n$ tiene que cumplirse hasta la precisión de la máquina.
Calcular $D_{1n}$ y $D_{2n}$ por caminos distintos y restarlos es una forma
barata de verificar que todo está bien.

## 3. Ecuaciones

**Las cuatro de Maxwell, completas:**

$$
\begin{aligned}
\nabla\cdot\mathbf{D} &= \rho_v
  && [\text{C}/\text{m}^3] \\[1ex]
\nabla\times\mathbf{E} &= -\frac{\partial\mathbf{B}}{\partial t}
  && [\text{V}/\text{m}^2] \\[1ex]
\nabla\cdot\mathbf{B} &= 0
  && [\text{T}/\text{m}] \\[1ex]
\nabla\times\mathbf{H} &= \mathbf{J} + \frac{\partial\mathbf{D}}{\partial t}
  && [\text{A}/\text{m}^2]
\end{aligned}
$$

**El argumento que obliga a incluir el término faltante:**

$$
\nabla\cdot(\nabla\times\mathbf{H}) \equiv 0
\quad\Longrightarrow\quad
\nabla\cdot\mathbf{J} + \frac{\partial}{\partial t}(\nabla\cdot\mathbf{D}) = 0
\quad\Longrightarrow\quad
\nabla\cdot\mathbf{J} = -\frac{\partial\rho_v}{\partial t}.
$$

Es decir: la ley de Ampère completa **contiene** a la ecuación de continuidad.

**Condición de borde normal:**

$$
D_{1n} = D_{2n}
\quad\Longrightarrow\quad
E_{2n} = \frac{\varepsilon_{r1}}{\varepsilon_{r2}} E_{1n},
\qquad
D_n = \varepsilon_0\varepsilon_r E_n .
$$

## 4. Qué significa físicamente

**El término faltante se agregó por consistencia, no por un experimento.**
Antes de Maxwell, la ley de Ampère se había verificado solo con corrientes
estacionarias, y nadie necesitaba
$\partial\mathbf{D}/\partial t$ para explicar nada. Maxwell lo agregó porque
la matemática no cerraba. Veinte años después se confirmó su consecuencia:
las ondas electromagnéticas que viajan a $c$.

**$\nabla\cdot\mathbf{B} = 0$ es la única que afirma una ausencia.** Ese cero
dice que no existen monopolos magnéticos: no hay "cargas magnéticas" sueltas.

**$E$ salta y $D$ no.** Al entrar a un material más permitivo, el campo
eléctrico normal baja porque el material se polariza y genera cargas ligadas
en la interfaz que lo apantallan. $\mathbf{D}$, en cambio, está definido para
"ver" solo la carga libre; como aquí no hay ninguna, pasa sin alterarse.

## 5. Parámetros modificables

Esta es la única celda que conviene editar. Cambie un valor, ejecute el notebook

```python
eps_r1 = 2.0        # permitividad relativa del medio 1
eps_r2 = 5.0        # permitividad relativa del medio 2
E1_normal = 10.0    # componente normal de E en el medio 1 [V/m]
```

## 6. Implementación

### 6.1 Revisión de las cuatro ecuaciones

Cada entrada indica cómo está escrita, si está completa y sus unidades.

```python
REVISION_MAXWELL = [
    (
        "div(D) = rho_v",
        "correcta",
        "C/m^3",
        "Ley de Gauss eléctrica; vale siempre.",
    ),
    (
        "curl(E) = -dB/dt",
        "correcta",
        "V/m^2",
        "Ley de Faraday; vale siempre.",
    ),
    (
        "div(B) = 0",
        "correcta",
        "T/m",
        "No existen monopolos magnéticos.",
    ),
    (
        "curl(H) = J",
        "incompleta",
        "A/m^2",
        "Falta dD/dt; solo vale si nada cambia en el tiempo.",
    ),
]

for ecuacion, estado, unidades, comentario in REVISION_MAXWELL:
    print(f"{ecuacion:<22} {estado:<12} [{unidades:<7}] {comentario}")
```

### 6.2 La condición de borde

```python
E2_normal = campo_normal_transmitido(E1_normal, eps_r1, eps_r2)

D1_normal = densidad_flujo_normal(E1_normal, eps_r1)
D2_normal = densidad_flujo_normal(E2_normal, eps_r2)

error_continuidad = D2_normal - D1_normal
```

## 7. Resultados numéricos

```python
tabla_resultados(
    [
        ("Campo normal en el medio 1", "E_1n", E1_normal, "V/m"),
        ("Campo normal en el medio 2", "E_2n", E2_normal, "V/m"),
        ("Razón entre campos", "E_2n/E_1n", E2_normal / E1_normal, "-"),
        ("Flujo normal en el medio 1", "D_1n", D1_normal, "C/m^2"),
        ("Flujo normal en el medio 2", "D_2n", D2_normal, "C/m^2"),
        ("Salto de D_n", "D_2n - D_1n", error_continuidad, "C/m^2"),
    ]
)
```

```python
print(f"Salto de D_n = {error_continuidad:.3e} C/m^2")
print(f"Razón esperada E_2n/E_1n = eps_r1/eps_r2 = {eps_r1 / eps_r2:.6f}")
assert abs(error_continuidad) < 1.0e-20, "D_n no es continuo"
print("La componente normal de D es continua.")
```

## 8. Visualización

Valores de $E_n$ y $D_n$ a cada lado de la interfaz, que está en $z = 0$. Se
ve de un vistazo cuál magnitud salta y cuál no.

```python
z = np.linspace(-1.0, 1.0, 400)
E_perfil = np.where(z < 0.0, E1_normal, E2_normal)
D_perfil = np.where(z < 0.0, D1_normal, D2_normal)

fig, (eje_E, eje_D) = plt.subplots(2, 1, sharex=True, figsize=(7.0, 5.5))

eje_E.plot(z, E_perfil, color="tab:blue")
eje_E.axvline(0.0, color="black", linestyle="--", label="interfaz")
eje_E.set_ylabel("E_n (V/m)")
eje_E.set_title(f"Interfaz eps_r1 = {eps_r1:.1f} | eps_r2 = {eps_r2:.1f}")
eje_E.set_ylim(0.0, 1.2 * max(E1_normal, E2_normal))
eje_E.legend()

eje_D.plot(z, D_perfil * 1.0e12, color="tab:orange")
eje_D.axvline(0.0, color="black", linestyle="--")
eje_D.set_xlabel("z (m)")
eje_D.set_ylabel("D_n (pC/m^2)")
eje_D.set_ylim(0.0, 1.2 * max(D1_normal, D2_normal) * 1.0e12)

fig.tight_layout()
plt.show()
```

## 9. Qué nos dicen los resultados

**La ecuación incompleta se delata sola.** No hizo falta ningún experimento
para descubrir que a la ley de Ampère le faltaba un término: bastó tomar la
divergencia y compararla con la conservación de la carga. Una inconsistencia
interna es suficiente para saber que algo está mal.

**El gráfico muestra dos comportamientos opuestos.** $E_n$ tiene un escalón en
$z = 0$: cae de 10.0 a 4.0 V/m al pasar a un medio 2.5 veces más permitivo.
$D_n$, en cambio, es una línea horizontal que cruza sin enterarse.

**El salto de $D_n$ es cero exacto.** No aproximadamente cero: la operación
$\varepsilon_2\cdot(\varepsilon_1/\varepsilon_2)E_{1n}$ devuelve el mismo
número de punto flotante que $\varepsilon_1 E_{1n}$. El `assert` lo verifica.
