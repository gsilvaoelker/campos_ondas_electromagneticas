## 2. Fundamentos teóricos y matemáticos

### 2.1 La auditoría de Maxwell

Las cuatro ecuaciones de Maxwell son el resumen completo del
electromagnetismo clásico. Un ejercicio útil consiste en revisarlas una por
una y decidir si están completas, en vez de aceptarlas como bloque.

El caso interesante es la ley de Ampère. La forma $\nabla\times\mathbf{H} =
\mathbf{J}$ es **incompleta**: solo vale cuando los campos no dependen del
tiempo. Tomando la divergencia de ambos lados se obtiene $\nabla\cdot
\mathbf{J} = 0$, porque la divergencia de un rotacional siempre es cero.
Pero la ecuación de continuidad dice que $\nabla\cdot\mathbf{J} =
-\partial\rho_v/\partial t$, que no es cero si la carga varía. La
contradicción se resuelve agregando la corriente de desplazamiento.

### 2.2 Consistencia dimensional

Verificar unidades es una herramienta de diagnóstico, no un trámite. Si los
dos lados de una ecuación no tienen las mismas unidades, la ecuación está
mal, sin necesidad de revisar la física.

### 2.3 La condición de borde como comprobación numérica

La continuidad de $D_n$ debe cumplirse hasta la precisión de la máquina.
Calcular $D_{1n}$ y $D_{2n}$ por caminos distintos y restarlos es una forma
de verificar que el resultado es correcto.

## 3. Ecuaciones relevantes

**Las cuatro ecuaciones de Maxwell, en su forma general:**

$$
\begin{aligned}
\nabla\cdot\mathbf{D} &= \rho_v
  && [\text{C}/\text{m}^3] \\[1ex]
\nabla\times\mathbf{E} &= -\frac{\partial\mathbf{B}}{\partial t}
  && [\text{V}/\text{m}^2] \\[1ex]
\nabla\cdot\mathbf{B} &= 0
  && [\text{T}/\text{m}] \\[1ex]
\nabla\times\mathbf{H} &= \mathbf{J} + \frac{\partial\mathbf{D}}{\partial t}
  && [\text{A}/\text{m}^2]
\end{aligned}
$$

**El argumento que obliga a incluir la corriente de desplazamiento:**

$$
\nabla\cdot(\nabla\times\mathbf{H}) \equiv 0
\quad\Longrightarrow\quad
\nabla\cdot\mathbf{J} + \frac{\partial}{\partial t}(\nabla\cdot\mathbf{D}) = 0
\quad\Longrightarrow\quad
\nabla\cdot\mathbf{J} = -\frac{\partial\rho_v}{\partial t}.
$$

Es decir: la ley de Ampère completa **contiene** la ecuación de continuidad.

**Condición de borde normal, sin carga libre superficial:**

$$
D_{1n} = D_{2n}
\quad\Longrightarrow\quad
E_{2n} = \frac{\varepsilon_{r1}}{\varepsilon_{r2}} E_{1n},
\qquad
D_n = \varepsilon_0\varepsilon_r E_n .
$$

## 4. Explicación física

**Por qué faltaba un término.** Antes de Maxwell, la ley de Ampère solo se
había verificado con corrientes estacionarias. El término
$\partial\mathbf{D}/\partial t$ no era necesario para explicar ningún
experimento conocido; se agregó por consistencia matemática. Su consecuencia
—la existencia de ondas electromagnéticas que viajan a $c$— se confirmó
experimentalmente veinte años después.

**$\nabla\cdot\mathbf{B} = 0$ no tiene fuente.** El cero del lado derecho
afirma que no existen monopolos magnéticos. Es la única de las cuatro
ecuaciones cuyo contenido es una **ausencia**.

**$E$ salta y $D$ no.** Al cruzar hacia un medio de mayor permitividad, el
campo eléctrico normal disminuye porque el medio se polariza y genera cargas
ligadas en la interfaz que apantallan parcialmente el campo. La densidad de
flujo $\mathbf{D}$, en cambio, está definida de modo que solo "ve" la carga
libre: como aquí no hay ninguna en la interfaz, pasa sin alterarse.

## 5. Parámetros modificables

Modifique **solo** esta celda y vuelva a ejecutar el notebook completo.

```python
eps_r1 = 2.0        # permitividad relativa del medio 1
eps_r2 = 5.0        # permitividad relativa del medio 2
E1_normal = 10.0    # componente normal de E en el medio 1 [V/m]
```

## 6. Implementación del código

### 6.1 Auditoría de las ecuaciones de Maxwell

Cada entrada indica la forma escrita, si está completa y sus unidades.

```python
AUDITORIA_MAXWELL = [
    (
        "div(D) = rho_v",
        "correcta",
        "C/m^3",
        "Ley de Gauss eléctrica; vale en todo régimen.",
    ),
    (
        "curl(E) = -dB/dt",
        "correcta",
        "V/m^2",
        "Ley de Faraday; vale en todo régimen.",
    ),
    (
        "div(B) = 0",
        "correcta",
        "T/m",
        "Ley de Gauss magnética; no existen monopolos.",
    ),
    (
        "curl(H) = J",
        "incompleta",
        "A/m^2",
        "Falta dD/dt; solo vale en régimen estacionario.",
    ),
]

for ecuacion, estado, unidades, comentario in AUDITORIA_MAXWELL:
    print(f"{ecuacion:<22} {estado:<12} [{unidades:<7}] {comentario}")
```

### 6.2 Condición de borde normal

```python
E2_normal = campo_normal_transmitido(E1_normal, eps_r1, eps_r2)

D1_normal = densidad_flujo_normal(E1_normal, eps_r1)
D2_normal = densidad_flujo_normal(E2_normal, eps_r2)

error_continuidad = D2_normal - D1_normal
```

## 7. Resultados numéricos

```python
tabla_resultados(
    [
        ("Campo normal en el medio 1", "E_1n", E1_normal, "V/m"),
        ("Campo normal en el medio 2", "E_2n", E2_normal, "V/m"),
        ("Razón entre campos", "E_2n/E_1n", E2_normal / E1_normal, "-"),
        ("Flujo normal en el medio 1", "D_1n", D1_normal, "C/m^2"),
        ("Flujo normal en el medio 2", "D_2n", D2_normal, "C/m^2"),
        ("Error de continuidad", "D_2n - D_1n", error_continuidad, "C/m^2"),
    ]
)
```

```python
print(f"Error de continuidad = {error_continuidad:.3e} C/m^2")
print(f"Razón esperada E_2n/E_1n = eps_r1/eps_r2 = {eps_r1 / eps_r2:.6f}")
assert abs(error_continuidad) < 1.0e-20, "D_n no es continuo"
print("La componente normal de D es continua.")
```

## 8. Visualización

Valores de $E_n$ y $D_n$ a cada lado de la interfaz, situada en $z = 0$.
La figura hace visible de un vistazo cuál magnitud salta y cuál no.

> *Nota:* esta figura no existe en el script original de la semana 6; se
> construye únicamente con las dos ecuaciones ya calculadas arriba y no
> introduce física nueva.

```python
z = np.linspace(-1.0, 1.0, 400)
E_perfil = np.where(z < 0.0, E1_normal, E2_normal)
D_perfil = np.where(z < 0.0, D1_normal, D2_normal)

fig, (ax_E, ax_D) = plt.subplots(2, 1, sharex=True, figsize=(7.0, 5.5))

ax_E.plot(z, E_perfil, color="tab:blue")
ax_E.axvline(0.0, color="black", linestyle="--", label="interfaz")
ax_E.set_ylabel("E_n (V/m)")
ax_E.set_title(f"Interfaz eps_r1 = {eps_r1:.1f} | eps_r2 = {eps_r2:.1f}")
ax_E.set_ylim(0.0, 1.2 * max(E1_normal, E2_normal))
ax_E.legend()

ax_D.plot(z, D_perfil * 1.0e12, color="tab:orange")
ax_D.axvline(0.0, color="black", linestyle="--")
ax_D.set_xlabel("z (m)")
ax_D.set_ylabel("D_n (pC/m^2)")
ax_D.set_ylim(0.0, 1.2 * max(D1_normal, D2_normal) * 1.0e12)

fig.tight_layout()
plt.show()
```

## 9. Interpretación física de los resultados

**La ley de Ampère incompleta se delata sola.** No hace falta un experimento
para descubrir que falta $\partial\mathbf{D}/\partial t$: basta tomar la
divergencia de la ecuación y comparar con la conservación de la carga. Una
inconsistencia matemática interna es suficiente para saber que la ecuación
está mal.

**El gráfico muestra dos comportamientos opuestos.** $E_n$ tiene un escalón
en $z = 0$: cae de 10.0 V/m a 4.0 V/m al pasar a un medio 2.5 veces más
permitivo. $D_n$, en cambio, es una línea horizontal que atraviesa la
interfaz sin inmutarse.

**El error de continuidad es cero exacto.** No es aproximadamente cero: la
operación $\varepsilon_2 \cdot (\varepsilon_1/\varepsilon_2) E_{1n}$ devuelve
el mismo número de punto flotante que $\varepsilon_1 E_{1n}$ en este caso.
El `assert` de la sección 7 lo verifica.
